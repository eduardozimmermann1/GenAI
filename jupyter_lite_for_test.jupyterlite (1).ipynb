```python
# Keep appending the code generated to this cell, or add more cells below this to execute in parts
import pandas as pd

# Specify the file path
file_path = "dataset.csv"

# Read the CSV file into a Pandas data frame
df = pd.read_csv(file_path)

# Assuming the first rows of the file are the headers, you don't need to specify any additional parameters

# Additional details:
# - The `pd.read_csv()` function is used to read a CSV file into a Pandas data frame.
# - By default, it assumes that the first row of the file contains the headers for the data.
# - If your file doesn't have headers, you can specify `header=None` as an additional parameter.
# - You can also specify other parameters, such as `sep` to specify the delimiter used in the file.
# - Make sure you have the Pandas library installed in your Python environment before running this code.
```

```python
import pandas as pd

# Verificar quais colunas têm valores ausentes
columns_with_missing_values = df.columns[df.isnull().any()]
print("Colunas com valores ausentes:", columns_with_missing_values.tolist())

# Tratar valores ausentes

# 1. Substituir valores ausentes de 'Screen_Size_cm' (categórica) pelo valor mais frequente (moda)
most_frequent_value = df['Screen_Size_cm'].mode()[0]
df['Screen_Size_cm'].fillna(most_frequent_value, inplace=True)

# 2. Substituir valores ausentes de 'Weight_kg' (contínua) pela média
mean_value = df['Weight_kg'].mean()
df['Weight_kg'].fillna(mean_value, inplace=True)

# Conferir se ainda restam valores nulos
print("\nValores ausentes após tratamento:")
print(df.isnull().sum())
```

```python
# 1. Verificar os tipos de dados atuais
print("Tipos de dados originais:\n")
print(df.dtypes)

# 2. Corrigir tipos de dados (se necessário)

# Exemplo: garantir que colunas numéricas sejam de fato numéricas
numeric_columns = ["Screen_Size_cm", "CPU_frequency", "RAM_GB", "Storage_GB_SSD", "Weight_kg", "Price"]

for col in numeric_columns:
    # remove caracteres não numéricos se existirem (ex: "2.5GHz" -> "2.5")
    df[col] = df[col].astype(str).str.extract(r'([\d\.]+)', expand=False)
    df[col] = pd.to_numeric(df[col], errors="coerce")

# 3. Conferir novamente os tipos
print("\nTipos de dados corrigidos:\n")
print(df.dtypes)

# 4. Mostrar estatísticas descritivas para confirmar que as colunas ficaram numéricas
print("\nEstatísticas descritivas das colunas numéricas:\n")
print(df[numeric_columns].describe())
```

```python
from sklearn.preprocessing import StandardScaler, MinMaxScaler

# Selecionar colunas numéricas (exceto talvez o ID "Unnamed: 0")
numeric_columns = ["Screen_Size_cm", "CPU_frequency", "RAM_GB", "Storage_GB_SSD", "Weight_kg", "Price"]

# Criar escaladores
scaler_standard = StandardScaler()
scaler_minmax = MinMaxScaler()

# Padronização (média = 0, desvio padrão = 1)
df_standardized = pd.DataFrame(
    scaler_standard.fit_transform(df[numeric_columns]),
    columns=[col + "_std" for col in numeric_columns]
)

# Normalização (valores entre 0 e 1)
df_normalized = pd.DataFrame(
    scaler_minmax.fit_transform(df[numeric_columns]),
    columns=[col + "_norm" for col in numeric_columns]
)

# Concatenar com o DataFrame original
df_scaled = pd.concat([df, df_standardized, df_normalized], axis=1)

# Verificar as primeiras linhas
df_scaled.head()
```

```python
# Selecionar colunas categóricas
categorical_columns = ["Manufacturer", "Screen", "GPU", "OS"]

# Aplicar One-Hot Encoding
df_encoded = pd.get_dummies(df_scaled, columns=categorical_columns, drop_first=True)

# Verificar resultado
print("Dimensões originais:", df_scaled.shape)
print("Dimensões após One-Hot Encoding:", df_encoded.shape)

# Mostrar primeiras linhas
df_encoded.head()
```

```python
import pandas as pd
from sklearn.preprocessing import StandardScaler, MinMaxScaler

# ======================================================
# 1. Importação do dataset
# ======================================================
file_path = "dataset.csv"  # já baixado no JupyterLite
df = pd.read_csv(file_path)

# ======================================================
# 2. Tratamento de valores ausentes
# ======================================================
# Substituir Screen_Size_cm (categórica) pela moda
most_frequent_value = df['Screen_Size_cm'].mode()[0]
df['Screen_Size_cm'] = df['Screen_Size_cm'].fillna(most_frequent_value)

# Substituir Weight_kg (contínua) pela média
mean_value = df['Weight_kg'].mean()
df['Weight_kg'] = df['Weight_kg'].fillna(mean_value)

# ======================================================
# 3. Correção de tipos de dados
# ======================================================
numeric_columns = ["Screen_Size_cm", "CPU_frequency", "RAM_GB", 
                   "Storage_GB_SSD", "Weight_kg", "Price"]

for col in numeric_columns:
    df[col] = pd.to_numeric(df[col], errors="coerce")

# ======================================================
# 4. Padronização e Normalização
# ======================================================
scaler_standard = StandardScaler()
scaler_minmax = MinMaxScaler()

# Padronização
df_standardized = pd.DataFrame(
    scaler_standard.fit_transform(df[numeric_columns]),
    columns=[col + "_std" for col in numeric_columns]
)

# Normalização
df_normalized = pd.DataFrame(
    scaler_minmax.fit_transform(df[numeric_columns]),
    columns=[col + "_norm" for col in numeric_columns]
)

# Concatenar ao dataset
df_scaled = pd.concat([df, df_standardized, df_normalized], axis=1)

# ======================================================
# 5. One-Hot Encoding de variáveis categóricas
# ======================================================
categorical_columns = ["Manufacturer", "Screen", "GPU", "OS"]
df_final = pd.get_dummies(df_scaled, columns=categorical_columns, drop_first=True)

# ======================================================
# Resultado final
# ======================================================
print("Dimensões originais:", df.shape)
print("Dimensões após transformações:", df_final.shape)
df_final.head()
```
